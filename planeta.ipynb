```python
Constante_Gravitacional = 6.67428e-11
Masa_Del_Sol = 1.9890e30
Masa_De_La_Tierra = 5.97219e24
Pos_Tierra_x =  1.496e11
Pos_Tierra_y =  0
Velocidad_inicial_x = 0
Velocidad_inicial_y = 29.783e3
Paso_Del_Tiempo =  3600*24
Tiempo_total = 3.154e7
N_pasos = int(Tiempo_total / Paso_Del_Tiempo)
Posinicial_Tierra = [Pos_Tierra_x, Pos_Tierra_y]
Velocidadinicial_Tierra = [Velocidad_inicial_x, Velocidad_inicial_y]
trayectoria = []
```

```python
def calcular_aceleracion(x, y):
  r = sqrt(x**2 + y**2)
  a_mag = -Constante_Gravitacional * Masa_Del_Sol / r**2
  ax = a_mag * (x / r)
  ay = a_mag * (y / r)
  return ax, ay
```

```python
def calcular_aceleracion(x, y):

    ax = -x
    ay = -y
    return ax, ay


x, y = 0.0, 0.0
vx, vy = 1.0, 1.0
dt = 0.01
N_pasos = 1000


trayectoria = []

for t in range(N_pasos):
    ax, ay = calcular_aceleracion(x, y)


    x_nuevo = x + vx * dt + 0.5 * ax * dt**2
    y_nuevo = y + vy * dt + 0.5 * ay * dt**2


    ax_nuevo, ay_nuevo = calcular_aceleracion(x_nuevo, y_nuevo)


    vx = vx + 0.5 * (ax + ax_nuevo) * dt
    vy = vy + 0.5 * (ay + ay_nuevo) * dt


    x, y = x_nuevo, y_nuevo

    trayectoria.append((x, y))

print(trayectoria[:10])
```

```python
import matplotlib.pyplot as plt
import numpy as np

def graficar_trayectoria(posiciones_x, posiciones_y, UA):
    x_ua = [x / UA for x in posiciones_x]
    y_ua = [y / UA for y in posiciones_y]

    plt.figure(figsize=(8, 8))

    plt.plot(x_ua, y_ua, 'b-', linewidth=1.5, label='Trayectoria Tierra')

    plt.plot(0, 0, 'yo', markersize=15, label='Sol', markerfacecolor='yellow', markeredgecolor='orange')

    plt.plot(x_ua[0], y_ua[0], 'go', markersize=8, label='Posición inicial')

    plt.xlabel('Posición X (UA)', fontsize=12)
    plt.ylabel('Posición Y (UA)', fontsize=12)
    plt.title('Trayectoria de la Tierra alrededor del Sol', fontsize=14)
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.axis('equal')
    plt.xlim(-1.5, 1.5)
    plt.ylim(-1.5, 1.5)

    plt.show()
```
